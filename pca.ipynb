```python
import pandas as pd
import numpy as np
#from sklearn.tree import DecisionTreeClassifier
#from sklearn.model_selection import train_test_split
#from lazypredict.Supervised import LazyClassifier
#from sklearn.neighbors import NearestCentroid
#from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
#import umap
#from sklearn.manifold import TSNE

df = pd.read_csv("term-deposit-marketing-2020.csv")

df["y"] = df["y"].map({"no": 0, "yes": 1})

binary_cols = ["housing", "loan", "default"]  # example

for col in binary_cols:
    df[col] = df[col].map({"no": 0, "yes": 1})

#df = df.drop(columns="contact")

df.replace('unknown', np.nan, inplace=True)


#binary_cols = ["housing", "loan", "default"]  # example

#for col in binary_cols:
#    df[col] = df[col].map({"no": 0, "yes": 1})

df_sub = df[df["y"] == 1].copy()
```

```python
cluster_features = [
    "age",
    "job",
    #"education",
    "marital",
    "default",
    "balance",
    "housing",
    "loan"
]
```

```python
X_cluster = df_sub[cluster_features].copy()
```

```python
X_cluster.info()
```

```python
pca = PCA()
```

```python
X_cluster = pd.get_dummies(
    X_cluster,
    columns=["job", "marital"],
    dtype=int
)
```

```python
X_cluster.shape
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_cluster)
```

```python
X_pca = pca.fit_transform(X_scaled)
```

```python
pca.explained_variance_ratio_
```

```python
X_pca[:, 0]   # PC1
```

```python
X_pca[:, 1]   # PC2
```

```python
plt.scatter(X_pca[:, 0], X_pca[:, 1])
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA: PC1 vs PC2")
plt.show()
```

```python
kmeans = KMeans(
    n_clusters=5,
    random_state=1234,
    n_init="auto"
)

clusters = kmeans.fit_predict(X_scaled)
```

```python
plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=clusters
)

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA with K-means Clusters")
plt.colorbar(label="K-means Cluster")
plt.show()
```

```python
np.cumsum(pca.explained_variance_ratio_)
```

```python
df_sub["cluster"] = clusters
X_cluster["cluster"] = df_sub["cluster"]
```
