```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
#from lazypredict.Supervised import LazyClassifier
from sklearn.neighbors import NearestCentroid
from sklearn.metrics import classification_report, confusion_matrix

df = pd.read_csv("term-deposit-marketing-2020.csv")

df.replace("unknown", np.nan, inplace=True)

df["y"] = df["y"].map({"no": 0, "yes": 1})

binary_cols = ["housing", "loan", "default"]

for col in binary_cols:
    df[col] = df[col].map({"no": 0, "yes": 1})

df["education"] = df["education"].map({"primary": 1, "secondary": 2, "tertiary": 3})

df_model1 = df.copy()
```

```python
drop_cols = [
    "contact",
    "duration",
    "campaign",
    "month",
    "day"
]

df_model1 = df_model1.drop(columns=drop_cols)
```

```python
df_model1 = pd.get_dummies(
    df_model1,
    columns=["job", "marital"],
    drop_first=True,
    dtype=int
)
```

```python
X = df_model1.drop("y", axis=1)
y = df_model1["y"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=1234,
    stratify=y
)
```

```python
tree = DecisionTreeClassifier(
    random_state=1234,
    criterion='log_loss',
    max_depth=27,
    min_samples_split=20,
    max_features=None,
    ccp_alpha=0.0003553899548391089,
    class_weight={0: 1, 1: 30},
    min_samples_leaf=6
)
```

```python
tree.fit(X_train, y_train)
```

```python
#predicted_subscribers=tree.predict(X_test)
predicted_subscribers=tree.predict(X)
```

```python
df['dt_prediction'] = predicted_subscribers
#df.loc[X_test.index, "dt_prediction"] = predicted_subscribers
filtered_df = df[df["dt_prediction"] == 1].copy()
```

```python
filtered_df["month"] = filtered_df["month"].map({"jan": 1, "feb": 2, "mar": 3, "apr": 4,  "may": 5,  "jun": 6, 
                                                 "jul": 7, "aug": 8, "sep": 9, "oct": 10, "nov": 11, "dec": 12})
```

```python
filtered_df = filtered_df.drop(columns="contact")
```

```python
filtered_df = pd.get_dummies(
    filtered_df,
    columns=["job", "marital"],
    drop_first=True,
    dtype=int
)
```

```python
filtered_df = filtered_df.drop(columns='education')
```

```python
X = filtered_df.drop("y", axis=1)
y = filtered_df["y"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=5678,
    stratify=y
)
```
